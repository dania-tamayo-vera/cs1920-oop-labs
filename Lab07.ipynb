## Searching Explained

_Searching_ is the action of accessing a dataset (which could be text, numerical information, or anything) and finding required information (or determining that it doesn't actually exist).

### Task 1 - Linear Search

A _linear_ search is exactly how you'd look for a file in a drawer of a filing cabinet that wasn't well-maintained - that is, the files were in no particular order.

You'd start at the beginning, and look at the file folder name - is that it?  No?  Next.

You'd repeat this action until you either _found_ the file you're looking for, or you reached the end of the files, and it _wasn't found_.

Normally, we are searching for the _first instance_ of a file, but if it's possible for multiple files to have the same name, you may want to retrieve the collection of files that match.  We're going to work with the simple assumption - first match is what we want.

In the data folder there is a file titled 'words_short_unsorted.txt' - that is the file we are using for this exercise.

You will write a function called *linear_search* that will take, as parameters, a list of words, and a target word. If the target word is found, return its index in the list - but if it is not found, return -1.  Thus, an integer result is expected.

We supply test code that will open the file, create the list of words, and will call your code to do the searching.

````python
# write your linear_search function below

def linear_search(data, target):
    for i in range(len(data)):
        if data[i] == target:
            return i
    return -1 # not found if we didn't already return


##### TEST CODE - DO NOT ALTER #####

words = []
with open('./data/words_short_unsorted.txt', 'r') as f:
    for line in f:
        words.append(line.strip())

targets = ["expertise","poor","club","avalanche"]

for target_word in targets:
    print(f'Linear search for "{target_word}": {linear_search(words, target_word)}')
````

Think about this for a minute - for a set of n elements, finding the last element takes n comparisons in _linear search_.  Finding that an element is not there also takes n comparisons - just that the last one doesn't actually find it, and you're then out of elements.

Finding the first element takes 1 comparison.  Finding the middle takes approximately n//2 comparisons (imprecise because the middle of an even-numbered list rounds down).

We've seen this before - where the amount of time it takes is directly proportional to the number of elements that must be processed, and we previously said this is $O(n)$ - Order n, or _linear time_.  That's why it's called a "linear search".  When data is unordered, a linear search is the only option we have available to us that is reliable.

But - what if the data were guaranteed to be sorted?  Can we do better?  YES WE CAN!

You will recall the "Guess My Number" game from CS-1910 where a random number between 1 and 100 is selected, and you have to guess it - and you do so by picking the midpoint of the range.  You then get either "Found it!", "Nope, higher!" or "Nope, lower!"  In the game you get 7 guesses, because $\lceil \log_2(100) \rceil = 7$

Using this logic, we expect $\lceil \log_2(10) \rceil$ to find an element in a sorted 10-element list in _at most_ 4 comparisons, and in a 1000-element list, $\lceil \log_2(1000) \rceil$ yields 10 comparisons, at most.  That's just over double the comparisons for 100 times the data!

In fact, one million items is just $\lceil \log_2(1000000) \rceil$ or 20 comparisons, and that's 100000 times the original 10-element data set!

We say that _binary search_ has a time complexity of $O(log(n))$ in its worst case.  That's way more efficient than $O(n)$.

_Why do we use the ceiling operator?_  You can't take a fraction of a guess, or in this context, perform a fraction of a comparison, so any fractional component must be rounded up (not fair-rounded).  That's what the ceiling $(\lceil)$ operator is showing.

## Task 2:  Binary Search (recursive)

Although this algorithm can easily be implemented without the use of recursion, we just covered recursion in our last lab, so it's a prime problem to employ it with.

Technique:
- find the midpoint of the data range
- Is that the target?  If so, YAY!  Return the target's location.
- If not, does the target come BEFORE the midpoint of the data range?
- - If so, then recursively repeat the search using the beginning of the range and one less than the midpoint as the end of the range.  In other words, you're disregarind the midpoint and anything above it, as the target cannot be in that range.
- If it's not before, then it must (by process of elimination) be AFTER.
- - Recursively repeat the search using the midpoint + 1 as the starting point, and the end of the range as the endpoint, thus eliminating the midpoint and anything before it.

If, at any time, the start and endpoints are reversed (that is, start > end), you've checked everyhing and have FAILED to find the item.  Return -1.

 Write a _binary_search_ function which takes, as parameters, a list, a start location, an end location, and a search target, and returns an integer (0 or more is a location, -1 is not found, as before).

````python
# write your binary_search function below, using recursion

def binary_search(data, start, end, word):
    if end < start:
        return -1

    mid_index = (start + end) // 2
    if data[mid_index] == word:
        return mid_index
    elif data[mid_index] > word:
        return binary_search(data, start, mid_index - 1, word)
    else:
        return binary_search(data, mid_index + 1, end, word)

##### TEST CODE - DO NOT ALTER #####
prefix = "./data/"
filenames = {"words_short_sorted.txt":("expertise","club","tense","avalanche"),
             "words_long_sorted.txt":("lethargy","ablaze","zips","avalanche")}

for filename in filenames:
    print(f'Binary searches on contents of "{filename}":\n')
    data = []
    with open(prefix+filename, "r") as f:
        for line in f:
            data.append(line.strip())
    for word in filenames[filename]:
        print(f'\tBinary search for "{word}": {binary_search(data, 0, len(data)-1, word)}')
    print()
````

## Sorting Explained

_Sorting_ is the action of accessing a dataset (which could be text, numerical information, or anything) and putting it in appropriate order.  This often involves files, and frequently creates a new sorted version of a file, or possibly overwrites the existing file with its sorted version.

You have learned in the labs how to add dunder comparator methods to an object - the most important of these is **__lt__** which is used by all built-in sorting options in python to determine order (to sort in reverse order, instead of using greather-than, we use not-less-than, it has the same effect.)  However, it is good practice to define _all six_ dunder comparators, because you may want to use custom code to do your sorting.

As stated above, there _are_ sort methods in Python - but other languages may not have them, and you may need to write your own.  Thus, you will write your own methods here.

Your textbook and lectures cover _insertion sort_, _merge sort_, and _quicksort_.  Rather than repeat these, we will look at two other sorts:  the inefficient (but often the first sort you'll encounter in many textbooks) _BubbleSort_, and the _SelectionSort_.  You will write both of these, and use our unsorted data files as your starting point.

Both of these methods rely on the _three bucket swap_ concept. Computers normally cannot handle exchanging the values of two variables in one operation.  Therefore, to swap the values of variables _a_ and _b_, we need a temporary variable _t_.  We then put the contents of a in t, the contents of b in a, and the contents of t in b.  Here's the code:

```python
t = a
a = b
b = t
```

Easy, isn't it?  But python allows this to be done in a single line of code (it still gets swapped like this internally, however):

```python
a,b = b,a
```

## Task 3 - Implement BubbleSort non-recursively

Given a list of unsorted data, sort it in ascending order as follows:

- compare the first two elements
- - if the elements are out of order, swap them
- compare the second (which may have just been swapped) and third elements
- - if they are out of order, swap them
- continue until you have compared the second-to-last element and the last element, and processed them accordingly

It is now a guarantee that the largest element is now at the end of the list; you will _never look at this element again_.

Now repeat the process from the start, but stop at the second-last position.
Then repeat with the third last position.
Continue this until the start and end are equal - that is, there's only one element remaining.  You are done, as the last element is in its proper place.

Write your **bubble_sort** function below.  It takes a list parameter, and returns the sorted version of that list.

````python
# write your bubble_sort function here

def bubble_sort(data):
    for last in range(len(data)-1,-1,-1):
        for current in range(0, last):
            if data[current] > data[current + 1]:
                data[current], data[current + 1] = data[current + 1], data[current] # swap!

##### TESTING CODE - DO NOT ALTER #####
def read_file(filename):
    data = []
    with open(filename, "r") as f:
        for line in f:
            data.append(line.strip())
    return data

print("Sorting short list.")
words = read_file("./data/words_short_unsorted.txt")
bubble_sort(words)
print(words)

print("Sorting long list.")
words = read_file("./data/words_long_unsorted.txt")
bubble_sort(words)
print("First five elements:")
print(words[:5])
print("Last five elements:")
print(words[-5:])
````

## Task 4 - Selection Sort

Selection sort is another basic sort type - and it's more efficient than BubbleSort, because bubblesort potentially swaps _every time_ a comparison is made.

Selection sort locates the desired element, then swaps it in one operation into the appropriate place.  This alone saves many unnecessary swaps.

The algorithm:

- assume the first element on the list is the largest by recording its index as such
- scan through the elements after the first element, to the end element, and if you find one bigger than the current largest, change the largest element's index to that element - all future comparisons will be against this element
- if you find another bigger, update the largest
- when you get to the end, you have the index of the largest element
- swap the largest and end elements
- reduce the end element index by one (never look at it again) and repeat the process until you have a single element left - and you are done.

Let's now code a **selection_sort** function to do the same task as before, but with this modified algorithm.

**NOTE:** we're demonstrating these as functions, but of course, they could be static methods or regular methods in a class, too.

````python
# write your non-recursive selection_sort function here

def selection_sort(data):
    for last in range(len(data)-1,-1,-1):
        largest = 0
        for current in range(1, last):
            if data[current] > data[largest]:
                largest = current
        data[largest],data[last] = data[last],data[largest] # swap!

##### TESTING CODE - DO NOT ALTER #####

def read_file(filename):
    data = []
    with open(filename, "r") as f:
        for line in f:
            data.append(line.strip())
    return data

print("Sorting short list.")
words = read_file("./data/words_short_unsorted.txt")
selection_sort(words)
print(words)

print("Sorting long list.")
words = read_file("./data/words_long_unsorted.txt")
selection_sort(words)
print("First five elements:")
print(words[:5])
print("Last five elements:")
print(words[-5:])
````
