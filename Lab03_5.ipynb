## Lab 03.5 - Inheritance and Polymorphism – Review

Inheritance and polymorphism are important concepts in object-oriented programming.  They allow us to create code that can apply to a wider range of object types, minimizing the amount of repetitive work, and increasing code quality by reducing potential sources of error.

First, let’s review some terminology:
- abstract class – a class that is functionally incomplete, and is used solely as an inheritance source and for polymorphic references.  This will normally be indicated with the use of method names that contain only the keyword pass (which means do nothing), and if it’s a method that ought to return something, it will return None as a default.  You are expected to implement such methods in your own code, overriding the functionally empty code from the abstract parent.  Thus, you write the exact same method header and simply add the code to do what the method’s documentation says it should do.
- instance variable – a variable declared, typically within __init__, with the self. prefix, making it class accessible.
- accessor – the formal name for a “getter” method.  Usually (but not always) takes no parameter (other than the required self), and returns a value of an instance variable.
- mutator – the formal name for a “setter” method.  Usually takes a parameter (besides self), and usually (but not always) does not return any value.
- dunder (aka magic) method – special pre-defined method names that are used for specific functions (like __str__) that Python will know how to use; 'dunder' refers to the double-underscore prefix and suffix used.
