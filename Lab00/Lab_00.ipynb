Remember, you need to open the output file as well as the input file - but the output file must be in write ('w') mode.  You cannot _print_ to a file - you have to _write_ to it, and you have to manually add a newline (\n) character to the end of each line.

Using f-strings will ensure that everything output is text, so if you've not been using them to solve the previous questions, now is a great time to bring that knowledge back in your brain!

````python
# insert your code BELOW this comment block, and above the final comment block shown below
# ensuring you make the changes outlined above

with open("output.txt","w") as output:
    with open("grades2.csv") as f:
        for line in f:
            line = line.strip()
            if line[0] == "#":
                continue
            data = line.split(",")
            name = data[0].strip()
            age = int(data[1].strip())
            attendance = int(data[2].strip())
            assignments = int(data[3].strip())
            midterm = int(data[4].strip())
            final = int(data[5].strip())
            comment = data[6].strip()
            grade = attendance/26*10 + assignments/40*20 + midterm/85*30 + final/120*40
            output.write(f"{name}: {grade:.2f}%\n")

# insert your code ABOVE this line
# this code tests whether your code did the job it was supposed to!

with open("output.txt") as result:
    for line in result:
        print(line.strip())
````
