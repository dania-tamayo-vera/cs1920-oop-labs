### Task Two - Dunder Comparator Methods

For one object to compare itself to another object _of the same type_, we use _dunder methods_.  "Dunder" refers to the double underscore (\_\_) characters that precede and follow the method name.  You already are familiar with two types of dunder methods, `__init__` and `__str__` which you met in the last lab.

There are six dunder comparators, which match the operations <, ==, >, <=, >=, and !=, and they are all of the same format:

`__dunder_method_name__(self, other):`

where _other_ is another instance of the same class.  Thus, you must use `isinstance` to check this first!

Then, you do the actual comparison based on the ivar needed.  For our purposes at this stage, we'll use the account holder's name, which is a string, and which you know how to compare using the conventional operators.

The six dunder methods are named:  `__lt__`, `__eq__`, `__gt__`, `__le__`, `__ge__`, and `__ne__`, respectively - and we're certain you can see where these two-letter names come from ... right?

Modify the BankAccount class (given) to include these six dunder methods.  Examine the test code carefully, and you'll see how each is activated.

## Section 2 - Using Static Methods and Static Variable

As mentioned, static variables are _not_ instance variables - they are variables that are shared across all instances of a given class (all objects created from a given class blueprint). What would this look like in terms of humans?

Let's say that you live with roommates - and you share all expenses including groceries, rather than having individual ownership of food items.  You might set up an app on your phones to maintain a shopping list - that way, if you're going to buy something, you don't buy something someone else already has gotten.  You devise a way (perhaps with last date bought) for items, and anything you input on your device shows up on your roommates' devices - and vice-versa.

That's a _static workspace_ in this context, and would behave like a set of _static variables_.  The app itself would be a set of _static methods_ which allow you to create, edit, delete, etc. things that are on the list.  This is the concept of _static_ within the context of object-oriented programming.

Now we are going to add a _static variable_ which will track what you want to compare by - and a _static method_ to set the basis for comparison.
