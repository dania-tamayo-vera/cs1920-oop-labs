## Lab06 - Recursion

We are now going to delve into a topic that may seem somewhat foreign to many of you - that of _recusion_.

The concept of recursion, on its surface, seems simple.  To solve a problem, take what is known from that problem, then solve a simplified version of the problem.  This is, of course, counter-intuitive.  We appear to be solving the problem by solving the problem, which of course is absurd.

Or is it?  The important part in this is the concept of the "simpler form" of the problem, and understanding where that relationship comes from.  That is key to understanding recursion.

All recursion has three basic requirements:
- one or more _recursive steps_ which invoke a simplified form of the original problem, and
- one or more _base cases_ (or _trivial conditions_) which have a concrete, known answer.
- the property that, in invoking a simplified form of the original problem, we _converge_ to a base case

So long as these properties hold, recursion will be a valid way to solve a given problem.  Without a recursive step, it's not recursion at all.  Without a base case, there's nothing to stop the recursion, and it becomes infinite - which, unlike a loop, will eventually crash the program.

One question that arises - when should you use recursion?  That's a difficult one to answer.  You should use it when the non-recursive solution is extremely complex, difficult or impossible to express in code, or is completely inflexible.  We'll look at a situation for that later.

Because recursion is, by definition, a repetition of steps, one could write a program which then writes the actual steps to a file while solving the problem, and that file would have no loops or other structures - just a huge sequence of steps.  That would solve THAT PARTICULAR INSTANCE of the problem, but would be very difficult to modify once "frozen" in non-recursive code.  There's not much real value in doing this.

Let's look at a recursive definition and its solution; we will then look at the non-recusive solution.

## Example 1A - Fibonacci Sequence via Recursion

The Fibonacci sequence mimics some things we see in nature (like the formation of a Nautilus' shell), and is defined by the mathematical expression:

$ F(n) = F(n-1) + F(n-2) $ - this is the _recurrence relation_.  Without some base case information, this is an infinite relationship.  We define the base cases as $ F(1) = 1 $ and F(0) = 1 $ - we are using 0 as the lowest base case, because Computer Scientists like to start things from 0.

Therefore, $ F(2) - F(1) + F(0) = 1 + 1 = 2 $ - thus, the third Fibonacci number is 2. By similar logic, $ F(3) = F(2) + F(1) = 2 + 1 = 3 $, and so on.

We will now code an example for you illustrating this principle.

````python
# Compute Fibonacci of n with the standard definition given above

def fibo_recusion(n):
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        return fibo_recusion(n-1) + fibo_recusion(n-2)

print(f"Fibonacci number 5 is {fibo_recusion(5)}")
print(f"Fibonacci number 6 is {fibo_recusion(6)}")
print(f"Fibonacci number 42 is {fibo_recusion(42)}")
````

**_What did you observe when you executed this code?_**

The first number appeared almost instantly on execution, and the second was perceptibly a bit longer in being generated - but it takes a lot longer to generate step 42.

Why is this?  To compute $F(42)$, you compute $F(41)$ and $F(40)$.  But then, to compute $F(41)$ you compute $F(40)$ and $F(39)$.

$F(40)$ gets computed _twice_.

This gets worse as we go along.  So, while the recursive relationship may seem natural for this problem, it is most definitely _not_ the most efficient way to express this algorithm.

Let's look at a non-recursive way to do this.

````python
# Compute Fibonacci of n with standard procedural programming (loops are linear)

def fibo_procedural(n):
    if n == 0:
        return 1
    elif n == 1:
        return 1
    else:
        sequence = [1,1] # this is what is expressed above
        for i in range(2,n):
            sequence.append(sequence[i-1] + sequence[i-2])

print(f"Fibonacci number 5 is {fibo_procedural(5)}")
print(f"Fibonacci number 6 is {fibo_procedural(6)}")
print(f"Fibonacci number 42 is {fibo_procedural(42)}")
````

_What did you note when you executed **THIS** code?_

Yes, it executed MUCH faster.  It also took much less memory, though you have no easy way to measure this in CoLab.

Why?

Whenever a recursive call is made, the _state_ of the function or method (yes, methods can also be recursive) has to be _saved_ - the parameter(s) passed, the values of local variables, and the point at which a new call was made - in a _stack frame_, which is something Python itself creates when any function or method is called.  With recursion, a lot of these frames get placed on the stack (which is why we taught you about stacks in the preceding lab).  When the function or method is complete, the result is held, the current stack frame is popped, and that result is passed to the previous stack frame, at the point where it called the function recursively.  This consumes _stack memory_ and takes additional time, as well.

Procedural programming does't generate a big stack of frames, so it can be a lot faster, and consumes much less memory.

_Why, then, use recursion at all?_

There are some problems which are _practically intractable_ (cannot effectively be solved) by standard procedural programming, or which are _practically inflexible_ when solved by such means.

An example of this is the classic Knight's Tour problem.  <https://en.wikipedia.org/wiki/Knight%27s_tour>

Solving for an 8x8 board takes quite a bit of time, and larger boards even moreso.  But you could run this algorith on a 5x5 board, or a 5x6 board, etc. With standard procedural programming, you'd have to create separate solutions for every variant of board size.  This is _practically inflexible_; the recursive solution allows us to specify the size of the board, and the solution algorithm is unchanged.  This one also qualifies as _practically intractable_, because for 8x8 or larger boards, the number of lines of code, if hand-written, could take years to complete.  There are approximately $4 \times 10^{51}$ possible move sequences that can be explored. This is using the standard solution (naïve backtracking) - there is something called Warnsdorff's Heuristic that can drastically reduce this, but that's the result of research to device a mathematical trick.

## Task 1 - Recursive Detection of a Palindrome in a String

A string is considered a palindrome if it reads the same left-to-right and right-to-left.  We normally eliminate spaces and punctuation, allowing only letters and numerals to be considered, and we usually ignore the case.

We will supply you with a function called _sanitize_ that takes a string, and returns the only-lowercase-letters-and-numbers format of that string.  We will pass you values that are sanitized, to a function called *is_palindrome* which takes a string and analyzes it to see if it's a palindrome, and returns a Boolean to indicate if it is or is not.

The word 'rattan' is not a palindrome, but 'deified' is. 'Redder' and 'Race Car' are once any  space is eliminated and it's reduced to all lowercase.  You will note that 'redder' has an even number of characters, and 'racecar' has an odd number - so there's no help in determining whether it has odd or even characters.

How do we check?  Look at the first and last characters - if they are NOT the same, it's not a palindrome, and we return __False__.  Otherwise, we check everything between the first and last characters, and get that result.

Trivial cases?  If there characters remaining is 1 or 0, it must be a palindrome - everything inspected before it was, and a single character is a reflection of itself (odd number of characters).  An empty string is similarly reflexively equivalent (nothing equals nothing!)

````python
# write your is_palindrome function below



##### GIVEN FUNCTION - SANITIZE - USED IN TEST CODE #####

def sanitize(phrase):
    new_phrase = ""
    for ch in phrase:
        if ch.isalnum():
            new_phrase += ch.lower()
    return new_phrase

##### TEST CODE - DO NOT ALTER #####

print(is_palindrome(sanitize("hello world"))) # False
print(is_palindrome(sanitize("rattan"))) # False
print(is_palindrome(sanitize("REDDER"))) # True
print(is_palindrome(sanitize("Race Car"))) # True
print(is_palindrome(sanitize("A Toyota"))) # True
print(is_palindrome(sanitize("A man, a plan, a canal: Panama"))) # True
print(is_palindrome(sanitize("I, madam, I MADE radio!  I dared!  Am I mad?  Am I?"))) # Surprise!
print(is_palindrome(sanitize("Recursion is hard.  :( "))) # False - wee what we did there?
````

_What did we learn?_

It's possible for us to use recursion to peel the characters like an onion, and as long as the layers remain equal, we'll find it's a palindrome.

Could we have done the same without recursion?  Of course.  In Python, we can use our slicing operations to traverse a string backwards.  Run the following demo; you should get the same output as above.

````python
def proc_is_palindrome(phrase):
    return phrase == phrase[::-1]

# We assume you completed and ran the code above, so the sanitize function is still available

##### TEST CODE - DO NOT ALTER #####

print(proc_is_palindrome(sanitize("hello world")))
print(proc_is_palindrome(sanitize("rattan")))
print(proc_is_palindrome(sanitize("REDDER")))
print(proc_is_palindrome(sanitize("Race Car")))
print(proc_is_palindrome(sanitize("A Toyota")))
print(proc_is_palindrome(sanitize("A man, a plan, a canal: Panama")))
print(proc_is_palindrome(sanitize("I, madam, I MADE radio!  I dared!  Am I mad?  Am I?")))
print(proc_is_palindrome(sanitize("Recursion is hard.  :( ")))
````
