# Lab 01 - Abstract Data Types, and Object Fundamentals

In this lab, you will begin exploring what _classes_ are, and how they aid us in encapsulating data and operations on that data into a single module.

Why do this?  Traditional programming languages like Pascal would have all of the data declared at the top of the program, but it may not get used until far, far down in the program file.  Since everything was in a single file, it became cumbersome to visualize and to work with.

Enter OOP - Object-Oriented Programming.  This paradigm, first seen in the language _SmallTalk_, allows us to _encapsulate_ data and operations within an object definition, keeping the data close to the operation code that will use it.  It then allows us to define operations the object can undertake (both internally to help modularize what the object does, but also externally, to limit how programmers can interact with the object).  We thus store data and related operations together, which makes logical sense, and makes programs far easier to read and interpret, and thus maintain.

In this lab, we will explore the concept of a "bank account" and how that would be encapsulated as a class.  Classes are blueprints for objects - once we define what an account is, and what can be done with it, we can create a variety of account _objects_ (that is, instance of the bank account _class_ blueprint), and have them store discrete information.
