## Lab05 - The Double-Ended Queue (deque) and Structures You Can Build

The deque (double-ended queue) is a structure that allows us to build other structures, using it as a basis.  THis means that we don't have to construct everything from scratch - at least, not in Python.

You'll learn about constructing data structures from scratch in CS-2910, but mostly in CS-2920.  Not all languages come with built-in building blocks, so as a Computer Scientist, you'll need to learn how some of these work so you can express them in languages that are less feature-rich.

First of all, in order to use the deque, you'll need to import it from a library:

```python
from collections import deque
```

This must happen anytime you want to use a deque in a given python file - so, any class file that requires the deque needs this import.

Let's test the deque with the codeblock shown below.  There's no "expected output", you're just going to run this, and observe what it does.

````python
from collections import deque

d = deque() # deque constructor
l = list() # using the list constructor to demonstrate similarity

# note how the deque isn't highlighted, because it's an import, but _list_ is highlighted, because it's a recognized python keyword.

s = "UPEI SMCS" # we're going to push these characters one-by-one into both the deque and the list
for ch in s:
    d.append(ch)
    l.append(ch)

# now, let's see what's in each:
print(f"Deque contents: {d}")
print(f"List contents: {l}")
````

Interesting!  So, the deque appears to internally be a list structure.  It looks identical to the list.  So - why use a deque?

A list has the primary operations of _append_ and _pop_ to add and remove elements at the end of the list, by default.  The deque also has these.  Make sure you've run the codeblock above, the following codeblock depends on that having been done.

````python
for i in range(4):
    print(f"Deque pop: {d.pop()}")
    print(f"List pop: {l.pop()}")
print()
print(f"Deque contents: {d}")
print(f"List contents: {l}")
````

Ok, so still no real difference. We're not convincing you, are we?  Ok, try this:

````python
for i in range(5): # there are 5 remaining elements at this point
    print(f"Deque popleft: {d.popleft()}")

for i in range(5): # so we can see the difference
    print(f"List popleft: {l.pop(0)}")
````

Ok, so deque has "popleft()" to pop from the left, but list doesn't - we just have to keep popping the leftmost element at index 0 to accomoplish the same thing.  A bit harder to read, but not terrible.

What about appending at the beginning, instead of at the end?

````python
s = "fantastic"
for ch in s: # both d and l are empty at this point
    d.appendleft(ch)
    l.insert(0, ch)

print(f"Deque contents: {d}")
print(f"List contents: {l}")
````

Again, we've managed to replicate the behaviour of the deque with the list.  However, I think you can see at this point, that the deque operations are easier to understand because they are via methods that are somewhat self-explanatory.  We had to use "insert" to accomplish what is essentially a left-sided append.  Remember, _code readability_ is an important concept, and something we want to promote! It will be easier for us, months from now, to understand our own intentions - and far easier for some other programmer who has to maintain our code after we got promoted (hopefully!) to actually do that task.

But there is a deeper reason, and it's something you may have seen already, but you definitely __will__ see it in your future!

When we use "insert(0, item)" to insert an item into a list, it actually appends a blank onto the end of the list, then moves, one-by-one, each element over, then assigns the 0-index element to the new data item.  This means that, when inserting into a list of n items, one operation makes the space, and then n move operations take place to shift each element over, then a final operation inserts the data item.  In Computer Science, we measure this via an "order expression" where we refer to the _biggest contributing factor_ in the operation, and leave out single operations. We know there are n+2 operations required for an insert, but we call this _O(n)_ in general, otherwise known as _linear time_.

The deque looks like it uses a list internally, but it's not actually a list, and it's managed completely differently.  It's actually written in C, which makes it superfast in execution, but more importantly, it uses something called a _linked list_ which you will learn in your second year courses. Although it "presents" as a list when printed, in reality, it's a completely different structure that simply allows an element to be added at the beginning in three operations (you'll have to trust us, but if you want an explanation, talk to your lab instructor!)

This is a fixed number of operations, and thus considered trivial when dealing with any real quantity of data - and so we call it O(1) - _constant time_.

**THE REALITY:**  Using a deque is inherently much more efficient than a list!

One of the things a stack can be used for is to convert an expression from _postfix notation_ (also known as _Reverse Polish Notation_ or RPN).  This notation has the notable characteristic that parentheses are _never_ required to override the order of operations, which is most definitely needed for _infix notation_ (the standard way we would write expressions mathematically.)

For example, consider this postfix expression:

` a b + c * = `

This is equivalent to the infix expression:

` ((a + b) * c) `

Yes, the outer set of parentheses is not necessary to include, but we're going to try and keep this simple - otherwise, we'd have to program awareness of precedence into any converter we might attempt to build.

Let's break down the process, but first, let's agree on the following:

- we shall limit operators to ['+','-','*','/','=']
- we shall not use the unary negative operator (again, to keep it simple)
- anything that is not an operator is considered an operand (like a value or variable)
- within our postfix expression, all elements are separated by whitespace
- you will not be asked to convert an invalid postfix expression - no error-checking is required

With these constraints, how do we interpret the first expression?

Well, since whitespace is a delimiter, we recognize that 'a' is not an operator (it is not _in_ the list of operators).  Therefore, we push it onto the stack.

We recognize that 'b' is also not an operator, and we therefore push it onto the stack.

We recognize that '+' _is_ an operator, so we pop what is on top of the stack as 'operand_2', and then pop the top of the stack _again_ to retrieve 'operand_1'.

We then form the expression '(' + operand_1 + operator + operand_2 + ')' remembering that we need spaces around the operator for clarity.  Perhaps an f-string might help with this . . .

Then, we push that expression back onto the stack.  '(a + b)' is now on the stack.

We then read the next element, which is 'c' - it's not an operator, so it goes on the stack.  The stack would now look like:

c
(a + b)

Now we read '*', an operator, so we pop the second operand [c] and the first operand [(a + b)], and form the new expression '((a + b) * c)' which we push back onto the stack.  It's now the only thing on the stack.

Finally, we read '=' - and we recognize that it's the 'finalization' operator, so we pop the stack and return that as the final expression.

Of course, this should work for any sort of expression.

It's not computing a result, just changing the format of the expression.
